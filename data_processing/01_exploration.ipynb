```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/RHL Season 10 Stats - OVERALL STATS.csv')
```

```python
df
```

```python
where_clause = df['GP'] >= 10
clean = df.loc[where_clause].copy()
```

```python
stat_columns = ['GPG', 'APG', 'SPG.1']
stat_means = clean[stat_columns].mean()
stat_stds = clean[stat_columns].std(ddof=0).replace(0, 1)
stat_means, stat_stds
```

```python
clean['attack'] = (75 + 10 * (clean['GPG'] - stat_means['GPG']) / stat_stds['GPG']).clip(0, 100)
clean['playmaker'] = (75 + 10 * (clean['APG'] - stat_means['APG']) / stat_stds['APG']).clip(0, 100)
clean['defence'] = (75 + 10 * (clean['SPG.1'] - stat_means['SPG.1']) / stat_stds['SPG.1']).clip(0, 100)
```

```python
clean['rating'] = clean['attack'] * 0.5 + clean['playmaker'] * 0.3 + clean['defence'] * 0.2
```

```python
clean.sort_values(by='rating', ascending=False)
```
